```python
def format_df(df, n=10_000):
    ...:     df = df.iloc[:n]
    ...:     df = df.drop_duplicates(subset=['latitude', 'longitude', 'date'])
    ...:     df = df.sort_values(['latitude', 'longitude', 'date'])
    ...:     index = pd.MultiIndex.from_arrays([df['latitude'], df['longitude'], df['date']])
    ...:     df.index = index
    ...:     latitudes = df.index.levels[0]
    ...:     longitudes = df.index.levels[1]
    ...:     times = df.index.levels[2]
    ...:     return latitudes, longitudes, times, df
```
